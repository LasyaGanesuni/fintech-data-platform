```python
from pyspark.sql import functions as F


def add_quarantine_metadata(
    df,
    reason,
    reason_column="_quarantine_reason"
):
    """
    Add standard quarantine metadata.
    """

    return (
        df
        .withColumn(
            reason_column,
            F.lit(reason)
        )
        .withColumn(
            "_quarantine_timestamp",
            F.current_timestamp()
        )
    )
```
